## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

# Task3(a):Missing values


```python
# T3a: Handle missing values

df = df.dropna(subset=["Brand", "Model"])

text_columns = df.select_dtypes(include="object").columns
df[text_columns] = df[text_columns].fillna("Unknown")

print("T3a: Shape after handling missing values:", df.shape)
print("T3a: Remaining missing values:")
print(df.isna().sum())
```

# Task 3(b) — Formatting

```python
# T3b: Standardize formatting

df["Registered City"] = df["Registered City"].str.strip().str.title()

df["Price"] = (
    df["Price"].astype(str)
    .str.replace("Rs", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

print("T3b: Sample cleaned cities:")
print(df["Registered City"].head())

print("T3b: Sample cleaned prices:")
print(df["Price"].head())
```

# Task 3(c) — Correct data types


```python
# T3c: Convert data types

df["Price"] = pd.to_numeric(df["Price"], errors="coerce").astype("int64")
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype("Int64")

print("T3c: Data types after conversion:")
print(df.dtypes)
```

# Task 3(d) — Spelling corrections

```python
# T3d: Correct spelling mistakes

df["Registered City"] = df["Registered City"].replace({
    "Sukkar": "Sukkur"
})

df["Model"] = df["Model"].str.replace(
    "Corrolla", "Corolla", regex=False
)

print("T3d: Corrected city names:")
print(df["Registered City"].value_counts().head())

print("T3d: Corrected Corolla models:")
print(df[df["Model"].str.contains("Corolla", case=False, na=False)]["Model"].head())
```

# Task 3(e) — Remove duplicate ads

```python
# T3e: Remove exact duplicates

duplicates_before = df.duplicated().sum()
df = df.drop_duplicates()

print("T3e: Duplicate rows found:", duplicates_before)
print("T3e: Shape after removing duplicates:", df.shape)
```

# Task 3(f) — Invalid values

```python
# T3f: Handle impossible KMs and fake prices

bad_km = (
    (df["KMs Driven"] == 1) &
    (df["Year"] == 1997) &
    (df["Model"].isin(["Prado", "Civic"]))
)

print("T3f: Invalid KM records found:", bad_km.sum())

df.loc[bad_km, "KMs Driven"] = np.nan

fake_price = df["Price"] == 87654321
print("T3f: Fake price records found:", fake_price.sum())

df = df[~fake_price]

print("T3f: Final shape:", df.shape)
```

# Task 3(g) — Final self-check

```python
# T3g: Final self-check

print("T3g: Final dataset shape:", df.shape)
print("T3g: Final data types:")
print(df.dtypes)
```
