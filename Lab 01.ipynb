Task 6: What is the main difference between a tuple and a list?

A list is mutable,which means we can change its elements after creating it.A tuple is immutable,which means we cannot change its element after creating it.Lists use square bracket [], while tuple uses parenthesis ().

Task 7: Function :grab the domain from "user@domain.com".

```python
def get_domain(email):
    return email.split('@')[1]

print("Q7:", get_domain("user@domain.com"))
```

Task 8: Function:return True if the word "dog" is in a string(case-insensitive)


```python
def find_dog(text): 
    return 'dog' in text.lower()
print("Q8:", find_dog("Is there a dog here?"))
print("Q8:", find_dog("Is there a cat here?"))
print("Q8:", find_dog("DOG is here"))
```

Task 9:Count how many times "dog" occurs in a string(case-insensitive)

```python
def count_dog(text): 
    return text.lower().count('dog')
print("Q9:", count_dog("This dog runs faster than the other dog"))
print("Q9:", count_dog("DOG dog Dog"))
```

Task 10: caught_speeding(speed,is_birthday)

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"


print("Q10:", caught_speeding(60, False))
print("Q10:", caught_speeding(75, False))
print("Q10:", caught_speeding(85, False))
print("Q10:", caught_speeding(85, True))
```
